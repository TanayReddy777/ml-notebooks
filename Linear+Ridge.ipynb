```python
import pandas as pd
df = pd.read_csv(r"C:\Users\tanay\Downloads\auto-mpg.csv")
df
```

```python
df.info
df.describe
df.isnull().sum()
```

```python
df.describe
```

```python
nums = ["cylinders","displacement","horsepower","weight","acceleration","model year","origin","mpg"]
for col in nums:
    df[col] = pd.to_numeric(df[col],errors="coerce")
```

```python
df.isnull().sum()
```

```python
df["horsepower"] = df["horsepower"].fillna(
    df["horsepower"].median()
)
```

```python
df.isnull().sum()
```

```python
df
```

```python
df = df.drop("car name",axis=1)
```

```python
x = df.drop("mpg",axis=1)
x.shape
```

```python
y = df["mpg"]
y.shape
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

lr = LinearRegression()
sc = StandardScaler()

x_train,x_test,y_train,y_test = train_test_split(x,y,random_state = 42, test_size = 0.2)
```

```python
x_train_std = sc.fit_transform(x_train)
x_test_std = sc.transform(x_test)
```

```python
lr.fit(x_train_std,y_train)
y_pred = lr.predict(x_test_std)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_absolute_error,root_mean_squared_error,r2_score,mean_squared_error
import numpy as np

print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("RMSE:", rmse)
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge,Lasso
ridge_model = Pipeline([
    ('poly',PolynomialFeatures(degree=5)),
    ('scaler',StandardScaler()),
    ('ridge',Ridge(alpha=1))
])
```

```python
ridge_model.fit(x_train,y_train)
ridge_train_pred = ridge_model.predict(x_train)
ridge_test_pred = ridge_model.predict(x_test)

print("\nRidge (degree = 5,alpha=1)")
print("Train Score: ",r2_score(y_train,ridge_train_pred))
print("Test score: ",r2_score(y_test,ridge_test_pred))
```

```python
from sklearn.preprocessing import StandardScaler,PolynomialFeatures
from sklearn.linear_model import Ridge,Lasso
from sklearn.pipeline import Pipeline
lasso_model = Pipeline([
    ('poly',PolynomialFeatures(degree=5)),
    ('scaler',StandardScaler()),
    ('lasso',Lasso(alpha=10,max_iter=10000))
])
```

```python
lasso_model.fit(x_train,y_train)
lasso_train_pred=lasso_model.predict(x_train)
lasso_test_pred=lasso_model.predict(x_test)
```

```python
print("\nLasso (degree = 5,alpha=10)")
print("Train Score: ",r2_score(y_train,lasso_train_pred))
print("Test score: ",r2_score(y_test,lasso_test_pred))
```
